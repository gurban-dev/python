```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('./HDI_Table.xlsx')
```

```python
# Drop rows in index positions 0, 1, 2, 3, 4, 5, and 6.
df.drop([0, 1, 2, 3, 4, 5, 6], inplace=True)
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
data_set = {
    'Countries': df['Table 1. Human Development Index and its components '],
    'Human Development Index (HDI)': df['Unnamed: 2'],
    'Life expectancy at birth': df['Unnamed: 4'],
    'Expected years of schooling': df['Unnamed: 6'],
    'Mean years of schooling': df['Unnamed: 8'],
    'Gross national income (GNI) per capita': df['Unnamed: 10']
}

df = pd.DataFrame(data=data_set)
```

Remove any row where one or more values is missing in the columns listed.

Missing values are often (NaN).

Rows that have values like ".." will not be removed.

```python
df.dropna(subset=[
    "Human Development Index (HDI)",
    "Life expectancy at birth",
    "Expected years of schooling",
    "Mean years of schooling",
    "Gross national income (GNI) per capita"
], inplace=True)
```

```python
for col in df.columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df = df.select_dtypes(include=['number'])
```

```python
df['Human Development Index (HDI)'] = df['Human Development Index (HDI)'].astype(float)
```

```python
df['Mean years of schooling'] = df['Mean years of schooling'].astype(float)
```

```python
df_plot = df[['Human Development Index (HDI)', 'Mean years of schooling']]
```

```python
df_plot.dtypes
```

```python
plt.scatter(
    df_plot['Human Development Index (HDI)'],
    df_plot['Mean years of schooling']
)
```
