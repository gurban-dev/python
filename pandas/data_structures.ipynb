At the core of Pandas are two data structures: <br>
The Series for working with array data <br>
The DataFrame for working with tabular data <br>

A Series is similar to a single column of data.

A DataFrame is similar to a sheet with rows and columns.

One way to create a Pandas DataFrame is by using a dictionary.

```python
level_of_unemployment = {
    # key: value
    # The keys are string literals.
    # The values are lists.
    "year": [2021, 2022, 2023, 2024, 2025],
    "unemployment_rate": [7.871, 7.303, 7.335, 7.4, 7.542]
}
```

```python
print(level_of_unemployment)
```

```python
level_of_unemployment
```

Create a Pandas DataFrame from the dictionary that 'level_of_unemployment' <br>
references.

```python
df = pd.DataFrame(level_of_unemployment)
```

```python
df
```

Inspet the data types of the DataFrame's columns.

```python
df.dtypes
```

Inspect the rows in one particular column.

```python
df["year"]
```

Plot the data in the DataFrame.

```python
ax = df.plot(
    x="year",
    y="unemployment_rate",
    title="Unemployment Rate in France (2021 - 2025)",
    ylabel="Unemployment Rate (%)",
    xlabel="Year",
    marker="o"
)

# Syntax:
# np.arange(start, stop, step)
ax.set_yticks(np.arange(2.0, 10.5, 0.5))

# Pandas uses the tools in matplotlib to draw the plot.
# By default, the numeric values on the x-axis are internally
# converted to floats.

# Controls where the ticks appear.
ax.set_xticks(df["year"])

# set_xticklabels() controls the text shown on the x-axis.
# astype() explicitly converts the data type of a DataFrame column.
ax.set_xticklabels(df["year"].astype(int))
```
