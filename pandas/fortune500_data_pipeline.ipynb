```python
import yfinance as yf
```

```python
# Pull historical stock data for multiple tickers.
microsoft = ["MSFT"]

# Passing an argument to a function without the name of the name
# parameter is called a "positional argument".
# E.g. microsoft

# Passing an argument to a function with the name of the name
# parameter is called a "keyword argument".
# E.g. tickers=microsoft

# Documentation for the download() function:
# https://ericpien.github.io/yfinance/reference/api/yfinance.download.html
data = yf.download(tickers=microsoft, start="1990-01-01")

print(data)
```

```python
import requests

# URL of the Wikipedia page containing the Fortune 500 companies list.
url = 'https://en.wikipedia.org/wiki/List_of_Fortune_500_companies'

# Set headers to mimic a browser request (helps prevent request blocks).
headers = {"User-Agent": "Mozilla/5.0"}

# Send a GET request to the URL and store the response.
response = requests.get(url, headers=headers)

# Use pandas to read all HTML tables from the page into a list of DataFrames.
tables = pd.read_html(response.text)

# pandas.read_html() returns a list; each item is a DataFrame corresponding
# to a table on the page.

# Using indexing allows you to grab only the first DataFrame:
# tables[0]

# Assuming the tickers are in the first table on the page (adjust index as needed).
df = tables[0]

# Print the full DataFrame to inspect its structure.
print(df)
```
