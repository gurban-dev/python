The Pandas Python library has a read_excel() function that accepts a string literal <br>
consisting of the path to the excel spreadsheet, and returns a DataFrame. <br>

Although read_excel() is being invoked with the dot operator, it is standalone, <br>
meaning that it is not called on an object. The dot operator is needed because <br>
it is defined inside the pandas module.

Supports the following file extensions read from a local file system or URL: <br>
xls, xlsx, xlsm, xlsb, odf, ods, and odt

Supports an option to read a single sheet or a list of sheets. <br>

Objective: <br>
Write a program that imports data from an excel spreadsheet and uses it to create a DataFrame.

Rows can be dropped from a dataset with the drop() method if they contain <br>
unwanted data: <br>
**data_frame_name.drop(index=data_frame_name.index[starting_row:ending_row], axis=0, inplace=True)**

Assigning 0 to the "axis" parameter means that rows will be dropped.

Assigning True to the "inplace" parameter will perform the operation in the <br>
original DataFrame rather than returning a copy of the updated DataFrame.

Construct a new DataFrame.

```python
# Dictionary.

# Each key in the dictionary is the name of a column.
# Each value is the data in the column.

data_set = {
    # column_name: column_data
    'Countries': df['Table 1. Human Development Index and its components '],
    'Human Development Index (HDI)': df['Unnamed: 2'],
    'Life expectancy at birth': df['Unnamed: 4'],
    'Expected years of schooling': df['Unnamed: 6'],
    'Mean years of schooling': df['Unnamed: 8'],
    'Gross national income (GNI) per capita': df['Unnamed: 10']
}

df = pd.DataFrame(data=data_set)
```

```python
df.head()
```

Remove any row where one or more values is missing in the columns
listed.

Missing values are often (NaN).

Rows that have values like ".." will not be removed.

```python
df.dropna(subset=[
    "Human Development Index (HDI)",
    "Life expectancy at birth",
    "Expected years of schooling",
    "Mean years of schooling",
    "Gross national income (GNI) per capita"
], inplace=True)
```
