A pandas Series is one-dimensional and therefore has only one axis. <br>
This axis is represented by the Series index and can be accessed through <br>
the index property.

```python
import pandas as pd
```

```python
import numpy as np
```

```python
# At this point, this is just a Python dictionary.
# A dictionary is made up of key-value pairs where each key
# corresponds to a value.
songs_dict = {
    # key: value
    'index': [0, 1, 2, 3],
    'data': [145, 142, 38, 13],
    'name': 'songs'
}
```

```python
songs_dict
```

Access the 'data' key in the 'series' dictionary.

```python
songs_dict['data']
```

Use the .index() method to find the index of a particular item.

The .index() method will return the index where the argument <br>
passed to it, is located.

```python
[0, 1, 2, 3].index(1)
```

```python
def get_value_from_dict(dictionary, item):
    # series['index'] will return [0, 1, 2, 3].
    value_idx = dictionary['index'].index(item)

    print("value_idx:", value_idx)

    return dictionary['data'][value_idx]
```

```python
get_value_from_dict(songs_dict, 1)
```

The .index() method accepts a piece of data and returns the index <br>
where that data is located in an iterable.

```python
[0, 1, 2, 3].index(1)
```

Get the index where the character "W" is located in the iterable "World".

```python
"World".index("W")
```

In a pandas Series constructor call, the order of arguments matters when <br>
they are passed positionally. However, when keyword arguments are used, <br>
the order does not matter because the parameters are matched by name.

```python
songs3 = pd.Series(
    data=[145, 142, 38, 13],
    name='counts',
    index=['Paul', 'John', 'George', 'Ringo']
)

songs3
```

dtype='object' is a generic Python data type container.

It means that the data is not made up of specialised NumPy types.

```python
songs3.index
```

The dtype that we see when we print a Series is the type of the <br>
values, not the index.

int64 is the standard NumPy integer data type. It means the data consists <br>
of integers, and each integer uses 64 bits (8 bytes) of memory.

```python
songs3.dtype
```

Although uncommon, a Series can contain values of different data types. <br>
In such cases, Pandas typically assigns the Series an object dtype.

When invoking the dtype property, it appears in the form dtype('O'), which <br>
indicates that the data is stored using the generic Python object type.

```python
class Foo:
    pass

mixed_data = pd.Series(
    data=['one', 1, 1.0, Foo()]
)
```

```python
# dtype('O')
mixed_data.dtype
```

```python
# dtype: object
mixed_data
```

When Pandas determines that a series holds numeric values but cannot <br>
find a number to represent an entry, it will use NaN.

The dtype of a Series will be float64 in the following context because
float64 supports NaN.

```python
nan_series = pd.Series(
    # Positional argument since the name of the parameter is not included.
    [2, np.nan],

    # Keyword argument since the name of the parameter is included.
    index=['Ono', 'Clapton']
)

nan_series
```

The .count() method, which counts the number of values in a series, ignores NaN.

1 is returned by the following since there is only one value in the list passed <br>
as the first argument that is not NaN.

```python
nan_series.count()
```

The 'size' property reveals the number of entries (values) in the Series.

It will take into account the NaN values.

```python
nan_series.size
```
