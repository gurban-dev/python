```python
import pandas as pd
```

```python
import numpy as np
```

```python
url = 'https://github.com/mattharrison/datasets/raw/master/data/vehicles.csv.zip'
```

Pandas read_csv() function can be used to read from a zip file <br>
provided that it only contains one file.

```python
df = pd.read_csv(url)
```

Inspect the Pandas DataFrame.

```python
df
```

Select a column from the DataFrame.

```python
city_mpg = df.city08
```

```python
city_mpg
```

Because the type of this series is int64, we know <br>
that none of the values are missing.

If a row is added to the Series that is NaN, the dtype <br>
becomes 'object'.

If len() is applied to the column named 'city_mpg', the number of rows <br>
in the column will be returned.

```python
len(city_mpg)
```

The .loc[] method is used to access a row or groups of rows from a <br>
DataFrame by the index.

In this case, the index label of the row being accessed is 41 144.

np.nan means "Not a Number" and nan is a pre-defined value in the NumPy <br>
library being what precedes .nan is np.

The square brackets that enclose np.nan indicate that a list is being <br>
assigned.

```python
city_mpg.loc[len(city_mpg)] = [np.nan]
```

```python
city_mpg
```
