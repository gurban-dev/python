```python
import pandas as pd
```

```python
import numpy as np
```

```python
url = 'https://github.com/mattharrison/datasets/raw/master/data/vehicles.csv.zip'
```

Pandas read_csv() function can be used to read from a zip file <br>
provided that it only contains one file.

```python
df = pd.read_csv(url)
```

```python
city_mpg = df.city08
```

```python
city_mpg
```

Because the type of this series is int64, we know <br>
that none of the values are missing.

If a row is added to the Series that is NaN, the dtype <br>
becomes 'object'.

```python
city_mpg.loc[len(city_mpg)] = [np.nan]
```

```python
city_mpg
```
