<h3>Practice: Categorising Temperatures with .apply() and .where()</h3>

````python
import pandas as pd

temperatures = pd.Series(
    [55, 107, 68, 110, 49, 73, 90, 30, 100, 83]
)
````

<h4>Part 1: Use .apply()</h4>

Create a function named classify_temp() that returns:
- "warm" for temperatures 70 or higher
- "cool" for temperatures below 70

Then use .apply() to classify every temperature.

````python
def classify_temp(temperature):
    if temperature >= 70:
        return "warm"
    else:
        return "cool"

temperatures.apply(classify_temp)
````

<h4>Part 2: Use .where()</h4>

Create a new Series where temperatures below 70 are replaced with "cool".

Temperatures 70 or higher should remain unchanged.

````python
temp_series = temperatures.where(temperatures >= 70, "cool")

temp_series
````

<h3>Part 3: Use .mask()</h3>

```text
Replace all temperatures above 105 with "Heat Warning".

Expected Output:
    temperature

2   Heat Warning
3   67
4   Heat Warning
```

<h3>Part 4: Use np.select()</h3>

```text
Create a column called "weather".

Rules:

Temperature        Label
Below 40           Cold
40-69              Mild
70-89              Warm
90+                Hot

Expected Output:
     temperature        weather
0             35           Cold
1             55           Mild
2             75           Warm
3             92            Hot
4             68           Mild
```

````python
import numpy as np

df = pd.DataFrame(
    {
        "temperature": [35, 55, 75, 92, 68]
    }
)

# Starter code:
df["weather"] = np.select(
    [
        df["temperature"] < 40,
        (df["temperature"] >= 40) & (df["temperature"] <= 69),
        (df["temperature"] >= 70) & (df["temperature"] <= 89),
    ],
    [
        "Cold",
        "Mild",
        "Warm"
    ],
    default="Hot"
)

df
````
