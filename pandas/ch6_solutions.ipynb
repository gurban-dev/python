4. Write an explanation of what the .add() method does.

The .add() method performs element-wise addition between two Pandas <br>
objects like Series or DataFrame.

Element-wise means a value-by-value addition at matching labels.

It's equivalent to using +, but with more control like fill_value.


.add(fill_value=0)

5. Fill in the missing code below to prevent the operation from returning NaN.

```python
# Does 's2' have a matching index 'a' for 's1'?

# Does 's1' have a matching index 'c' for 's2'?

s1 = pd.Series([1, 2], index=['a', 'b'])

s2 = pd.Series([10, 20], index=['b', 'c'])
```

```python
s1_s2_combination = s1.add(s2)

print("s1_s2_combination:\n", s1_s2_combination, sep="")
```

```python
# When a corresponding index (an index with the same name) doesn't
# exist in the other Series, fill_value=0 provides a default value
# of 0 (zero) for the missing index.
s1_s2_combination = s1.add(s2, fill_value=0)

print("s1_s2_combination:\n", s1_s2_combination, sep="")
```
