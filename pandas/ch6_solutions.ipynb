```python
import pandas as pd
```

```python
hdi = pd.Series(
    data=[0.972, 0.970, 0.970, 0.962, 0.959],
    index=["Iceland", "Norway", "Switzerland", "Denmark", "Germany"]
)
```

1. Add the numeric series to itself.

```python
hdi + hdi
```

2. Add 10 to the numeric series.

```python
hdi + 10
```

3. Add the numeric series to itself using the .add() method.

   Also, write out how you would add the series to itself when
   the indices do not perfectly match.

```python
hdi.add(hdi)

# The result here is the same as hdi + hdi.

# Adding the following way becomes useful when indices do not
# perfectly match:
hdi.add(hdi, fill_value=0)
```

4. Read the documentation for the .add() method.

According to the Pandas documentation, the .add() method:

- Performs element-wise addition
- Aligns values based on the index
- Supports a fill_value parameter to handle missing data
- Is part of a family of flexible arithmetic methods (add, sub, mul, div, etc.)

<h3>Key takeaway</h3>

- Use + for simple cases
- Use .add() when you want more control, especially with missing data or method chaining
