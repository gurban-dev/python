# Understanding Pandas Data Types and Memory Usage

Part 1: Create a numeric Series and inspect memory usage.

```python
import pandas as pd

int64_series = pd.Series([10, 20, 30, 40, 50])

print(int64_series)
print()

# Print the dtype of the Series.
print("int64_series.dtype:", int64_series.dtype)
print()

# Print the number of bytes used by the underlying data.
print("Memory usage (bytes):", int64_series.nbytes)
print()

# Explanation:
# By default, pandas usually stores integers as int64.
# An int64 uses 64 bits (8 bytes) per value.

# We have 5 values:
# 5 × 8 bytes = 40 bytes

# Therefore .nbytes returns 40.
```

Part 2: Convert to smaller integer types

```python
# Convert the Series to int8.
int8_series = int64_series.astype("int8")

print("int8_series dtype:", int8_series.dtype)
print()

print("int8_series memory usage (bytes):", int8_series.nbytes)
print()

# Compare with the original Series.
print("int64_series memory:", int64_series.nbytes)
print("int8_series memory:", int8_series.nbytes)
print()

# Explanation:

# int64:
#   64 bits = 8 bytes per value

# int8:
#   8 bits = 1 byte per value

# For 5 values:
# int64:
#   5 × 8 = 40 bytes

# int8:
#   5 × 1 = 5 bytes

# Smaller dtypes require less memory.
```

Part 3: Trigger an integer overflow problem

```python
large_number = pd.Series([128])

print("Original:")
print(large_number)
print()

# Attempt to convert to int8.
overflow_result = large_number.astype("int8")

print("After int8 conversion:")
print(overflow_result)
print()

# Explanation:

# int8 can store only:
# -128 through 127

# because it has only 8 bits available.

# The value 128 is outside this range.

# Pandas/Numpy do not raise an error here.

# Instead, the value wraps to the next representable signed
# 8-bit value which is -128.

# 128 becomes -128.

# This is called integer overflow.

# Range summary:
# int8:
#   -128 to 127

# int16:
#   -32,768 to 32,767

# int32:
#   -2,147,483,648 to 2,147,483,647

# int64:
#   Very large integer range
```

Part 4: Use convert_dtypes()

```python
df = pd.DataFrame({
    "age": [25, 30, 35],
    "name": ["Alice", "Bob", "Charlie"],
    "is_student": [True, False, True]
})

print("df.dtypes:\n", df.dtypes, sep="")

# Attempting to put a missing value into a column with a int dtype
# will coerce the column into having a float dtype.
df.loc[2, "age"] = pd.NA

# str can support missing values.
df.loc[2, "name"] = pd.NA

print("\ndf.dtypes:\n", df.dtypes, sep="")

# Attempt to insert a missing value into a column with a bool dtype
# raises a TypeError.
# df.loc[2, "is_student"] = pd.NA

print("\ndf:\n", df, sep="")
      
# Convert columns to pandas' nullable dtypes.
# A nullable data type allows a column to contain a missing value
# (pd.NA).
df_converted = df.convert_dtypes()

# Attempt to insert a missing value without a nullable data type.
df_converted.loc[2, "is_student"] = pd.NA

print("\ndf_converted.dtypes():\n", df_converted.dtypes, sep="")

print("\ndf_converted:\n", df_converted, sep="")

# Explanation:
# convert_dtypes() attempts to replace standard NumPy
# dtypes with newer pandas extension dtypes.

# Typical conversions:
# int64     -> Int64
# object    -> string
# bool      -> boolean

# These dtypes support missing values (pd.NA)
# more consistently than traditional NumPy types.
```
