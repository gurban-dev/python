<h3>Part 1: Transforming and Reshaping Data with Pandas with .apply()</h3>

Suppose there is a column of data.

How can a modified version of this column be created?

````python
scores = pd.Series([85, 82, 91, 73, 58])
````

Maybe the following is needed instead:

["low, "high", "high", "high", "low"]

If the data was [65, 82, 90, 73, 58], instead it become:

["D", "B", "A", "C", "F"]

These are all examples of manipulating a Series.

.apply() runs a function on every value.

````python
import pandas as pd

scores_series = pd.Series([65, 82, 90, 73, 58])

def classify(score):
    if score >= 70:
        return "high"
    else:
        return "low"

# Modifies the existing Series.
scores_series.apply(classify)
````

<h4>Mental Model</h4>

Imagine pandas doing this one value at a time:

````python
result = []

for score in scores_series:
    result.append(classify(score))

result
````

<h4>Why is .apply() often discouraged?</h4>

.apply() is a built-in method that belongs to the Pandas library.

It is discouraged because it is usually slower than Pandas' built-in <br>
vectorised operations, which are optimized for performance.

If a Series has 1 000 000 rows, then .apply() calls a function <br>
one million times which is computationally expensive.

It incurs a significant function call overhead and is typically <br>
much slower than using pandas' operations.

Can pandas do this without calling a function repeatedly? <br>
Usually the answer is yes.

<h3>Part 2: .where()</h3>

series.where(condition, other=value)

Keep the items where the condition evaluates to True, otherwise <br>
replace each item with the value passed as the second argument to <br>
the Series constructor.

````python
scores_series = pd.Series([65, 82, 91, 73, 58])

# Keep scores that are greater than or equal to 70, and replace
# scores that are less than 70 with "low".
scores_series.where(scores_series >= 70, "low")

# If the condition evaluates to True, the original value is preserved.

# If the condition evaluates to False, the original value is replaced
# with "low".
````

<h3>Part 3: .mask()</h3>

series.mark(condition, other=value)

Replace each item with 'value' where the condition evaluates to True.

.mask() is the opposite of .where().

.mask() probably will be rarer that .where().

````python
scores_series.mask(scores_series < 70, "low")
````

<h3>Part 4: np.select()</h3>

```text
This resembles if/elif/else.

The difference between .where() and np.select() is that .where() only allows one condition whereas with np.select() multiple are permissible.

Suppose:
score >= 90 -> A
score >= 80 -> B
otherwise -> C

Syntax: <br>
np.select(
    list_of_conditions,
    list_of_choices,
    default=0
)

list_of_conditions is a list of Boolean conditions.

list_of_choices is a list of values corresponding to each condition.

default is the value used when none of the conditions are True.
```

````python
import numpy as np

np.select(
    [
        scores_series >= 90,
        scores_series >= 80
    ],
    [
        "A",
        "B"
    ],
    default="C"
)
````

<h3>Part 5: Missing Data</h3>

NaN means "Not a number" and indicates that there is a missing value.

It does not mean 0 (zero).

````python
import pandas as pd
import numpy as np

temps = pd.Series([70, 72, np.nan, 75])
````

<h4>Detecting Missing Values</h4>

The .isna() method will replace each item with True if that item <br>
is NaN, and False otherwise.

````python
temps.isna()
````

<h4>Count the Missing Values</h4>

````python
temps.isna().sum()
````

<h3>Part 6: .fillna()</h3>

The .fillna() method replaces missing values.

````python
# Replace all missing values with 0.0.
temps.fillna(0)
````

````python
# Replace all missing values with 0.
temps.fillna(0).astype(int)
````

<h4>More realistc:</h4>

````python
# Replace missing values with the median/midpoint.
temps.fillna(temps.median())
````

````python
# Replace missing values with the median/average.
temps.fillna(temps.mean())
````

An outlier is an extreme value.

Question:
Is the mean or median usually better for data with outliers?

Answer:
The median is often preferred over the mean when the dataset contains
extreme values because the median is resistant to unusually large or
unusually small observations.

Consider the incomes of five people:
$40 000
$45 000
$50 000
$55 000
$ 2 000 000

The mean would be computed by adding all values and dividing by 5.

(40 000 + 45 000 + 50 000 + 55 000 + 2 000 000) / 5 -> 438 000

The mean is $438 000

The median would be calculated by sorting the values in ascending
order and taking the middle value.

$40 000
$45 000
$50 000 <- middle value
$55 000
$ 2 000 000

The median is $50 000.

Most people in the dataset earn between $40 000 and $55 000.

The mean of $438 000 suggests a typical income that 80% of the group
doesn't actually earn.

The median of $50 000 is much closer to what a typical person earns.
