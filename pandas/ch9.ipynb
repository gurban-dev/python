<h3>Part 2: .apply()</h3>

.apply() runs a function on every value.

````python
import pandas as pd

scores_series = pd.Series([65, 82, 90, 73, 58])

def classify(score):
    if score >= 70:
        return "high"
    else:
        return "low"

# Modifies the existing Series.
scores_series.apply(classify)
````

<h4>Mental Model</h4>

Imagine pandas doing this one value at a time:

````python
result = []

for score in scores_series:
    result.append(classify(score))

result
````

<h4>Why is .apply() often discouraged?</h4>

.apply() is a built-in method that belongs to the Pandas library.

It is discouraged because Python is slow compared to Pandas' <br>
optimised operations.

If a Series has 1 000 000 rows, then .apply() calls a function <br>
one million times which is computationally expensive.

It incurs a significant function call overhead and is typically <br>
much slower than using pandas' operations.

Can pandas do this without calling a function repeatedly? <br>
Usually the answer is yes.

<h3>Part 3: .where()</h3>

series.where(condition, other=value)

Keep the items where the condition evaluates to True, otherwise <br>
replace with value.

````python
scores_series = pd.Series([65, 82, 91, 73, 58])

# Keep scores that are greater than or equal to 70, and replace
# scores that are less than 70 with "low".
scores_series.where(scores_series >= 70, "low")

# If the condition evaluates to True, the original value is preserved.

# If the condition evaluates to False, the original value is replaced
# with "low".
````

<h3>Part 4: .mask()</h3>

series.mark(condition, other=value)

Replace each item with 'value' where the condition evaluates to True.

.mask() is the opposite of .where().

.mask() probably will be rarer that .where().

````python
scores_series.mask(scores_series < 70, "low")
````

<h3>Part 5: np.select()</h3>

```text
This resembles if/elif/else.

The difference between .where() and np.select() is that .where() only allows one condition whereas with np.select() multiple are permissible.

Suppose:
score >= 90 -> A
score >= 80 -> B
otherwise -> C

Syntax: <br>
np.select(
    list_of_conditions,
    list_of_choices,
    default=0
)

list_of_conditions is a list of Boolean conditions.

list_of_choices is a list of values corresponding to each condition.

default is the value used when none of the conditions are True.
```

````python
import numpy as np

np.select(
    [
        scores_series >= 90,
        scores_series >= 80
    ],
    [
        "A",
        "B"
    ],
    default="C"
)
````
