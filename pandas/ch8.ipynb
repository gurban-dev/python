<h5>1. Why Data Type Conversion Matters</h5>

Examples:
- int64
- float64
- object
- category
- string

Choosing the correct data type matters because it affects:
- Memory usage
- Available operations
- Handling missing values

Example: <br>
A column stored as int64 uses more memory than Int16 if the <br>
values are small enough to fit inside a 16-but integer.

<h5>2. Use .astype() and .convert_dtypes() correctly.</h5>

Pandas introduced .convert_dtypes() to automatically convert <br>
columns into more appropriate Pandas data types.

```python
import pandas as pd

num_series = pd.Series(
    [1, 2, None],
    dtype='Int64'
)

# With int64 set as the dtype, including None would raise a
# TypeError exception.

print("num_series:\n", num_series, sep="")

# int64 is a NumPy integer type.
# Int64 is a Pandas nullable integer type that supports pd.NA.

# Traditional NumPy integers cannot store missing values.
# Pandas nullable integers can.
```

Explicit Conversion with .astype()

Syntax: <br>
<name_of_the_series>.astype(dtype)

Sometimes it is appropriate to use smaller types because they <br>
consume less memory space.

int64 uses 8 bytes. <br>
int32 uses 4 bytes. <br>
int16 uses 2 bytes. <br>
int8 uses 1 byte.

.astype()

.astype() is used when you want explicit control over the target data type.

You manually specify the exact type you want.

```python
num_series.astype('Int16')
```

.convert_dtypes()

.convert_dtypes() attempts to automatically convert data into modern <br>
pandas types that support missing values (pd.NA).

```python
num_series.convert_dtypes()
```

.astype() converts data to a specific type that you explicitly choose, <br>
giving you full control over the result.

.convert_dtypes() automatically attempts to convert columns into the <br>
most appropriate pandas nullable types that support missing values <br>
(pd.NA).

<h5>3. Reduce memory usage with smaller numeric types.</h5>

Every numeric type has limits.

```python
import numpy as np

np.iinfo('int8')
```

```python
pd.Series([127]).astype('Int8')
```

```python
# Raises a TypeError exception.
# pd.Series([128]).astype('Int8')
```

```python
# With NumPy int8 types, values outside the valid range do not
# raise an error. Instead, they overflow and wrap around using
# two's complement binary arithmetic.

# Since int8 only supports values from -128 to 127, the value
# 128 wraps around after exceeding the maximum limit:
# 128 - 256 -> -128
pd.Series([128]).astype('int8')
```
