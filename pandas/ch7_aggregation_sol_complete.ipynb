3. What is the average (mean) median wage across all states?

   What is the highest median wage, and which state does it belong to?
   
   What is the lowest median wage, and which state does it belong to?
   
   How many total entries are in the dataset?
   
   How many unique wage values are there?

````python
# Average/mean wage of all the median wages.
mean_wage = median_wages_by_state_descending.mean()

print("mean_wage:", mean_wage)

# Highest median wage.
highest_wage = median_wages_by_state_descending.max()

print("\nhighest_wage:", highest_wage)

# Find the name of the state with the highest median wage.
highest_wage_state = median_wages_by_state_descending.idxmax()

print("\nhighest_wage_state:", highest_wage_state)

# Lowest median wage.
lowest_wage = median_wages_by_state_descending.min()

print("\nlowest_wage:", lowest_wage)

# Find the name of the state with the lowest median wage.
lowest_wage_state = median_wages_by_state_descending.idxmin()

print("\nlowest_wage_state:", lowest_wage_state)

# Get the total number of entries.
total_no_of_entries = median_wages_by_state_descending.size

print("\ntotal_no_of_entriess:", total_no_of_entries)

# Find the number of unique wage values.
no_of_unique_wages = median_wages_by_state_descending.nunique()

print("\nno_of_unique_wages:", no_of_unique_wages)
````

4. Find the following quantiles: <br>
   10th percentile

   50th percentile (median)

   90th percentile

   What does the difference between the 10th and 90th percentile <br>
   tell you about wage spread?

````python
# A percentile is any position out of 100.

# A quantile is a general concept of splitting data into equal parts.

# The data can be split into 4 quantiles or 5 quantiles.
# All of the 4 quantiles must be equal in size. The same applies
# to the 5 quantiles.

# Find the 10th, 50th and 90th percentiles.
quantiles = median_wages_by_state_descending.quantile([0.1, 0.5, 0.9])

print("quantiles:\n", quantiles, sep="")

# .loc[] is an indexer invoked on Pandas Series for the purpose of
# finding a value that corresponds to a particular index label.

# Obtain the value from this Pandas Series whose index label is 0.1.
q10 = quantiles.loc[0.1]

q50 = quantiles.loc[0.5]
q90 = quantiles.loc[0.9]

print("\n10th percentile:", q10)
print("50th percentile:", q50)
print("90th percentile:", q90)

# Compute the median wage gap between 10th and 90th percentile.
median_wage_gap_10_90 = q90 - q10

print("\n10th-90th percentile median wage gap:", median_wage_gap_10_90)
````

5. How many states have a median wage above $50 000?

   What percentage of states have a median wage above $50 000?

   How many states fall below $45 000, and what percentage is that?

````python
# Boolean Aggregation

# States above $50 000
above_50k_count = (median_wages_by_state_descending > 50_000).sum()
above_50k_percent = (median_wages_by_state_descending > 50_000).mean() * 100

print("States above $50K:", above_50k_count)
print("Percent above $50K:", above_50k_percent)

# States below $45 000
below_45k_count = (median_wages_by_state_descending < 45_000).sum()
below_45k_percent = (median_wages_by_state_descending < 45_000).mean() * 100

print("States below $45K:", below_45k_count)
print(f"Percent below $45K: {below_45k_percent:.2f}")
````

8. Create a custom aggregation: <br>
   ```python
   def wage_range(s): <br>
       return s.max() - s.min()
   ```

  Add this to .agg() and answer: <br>
- What is the total wage spread across states?

- Is that spread large or small relative to the mean?

````python
# Custom Aggregation
def wage_range(s):
    return s.max() - s.min()

agg_with_range = median_wages_by_state_descending.agg([
    'mean',
    'min',
    'max',
    'nunique',
    'std',
    wage_range
])

print("\nAggregations with wage range:")
print(agg_with_range)

total_spread = wage_range(median_wages_by_state_descending)
mean_wage = median_wages_by_state_descending.mean()

print("\nTotal wage spread:", total_spread)
print("Mean wage:", mean_wage)
print(f"Spread relative to mean: {total_spread / mean_wage:.2f}")
````
