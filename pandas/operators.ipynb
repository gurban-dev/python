```python
s1 = pd.Series([10, 20, 30], index=[1, 2, 2])

s2 = pd.Series([35, 44, 53], index=[2, 2, 4], name='s2')
```

```python
s1
```

```python
s2
```

Adding Series 's1' and 's2' demonstrates why having duplicate <br>
index labels should be avoided.

When a label no longer uniquely identifies a value, it becomes unclear <br>
what data s1[2] returns. It returns multiple values, not one.

20 + 35 = 55

20 + 44 = 64

30 + 35 = 65

30 + 44 = 74

```python
s1 + s2
```

When adding two Pandas Series, they align by index.

If an index exists in one Series but not the other, the result
would normally be NaN as seen with the above output.

fill_value=0 means that Pandas treats missing values as 0 before
adding. If a corresponding value is missing in either Series,
pretend it is 0 during the operation.

```python
s1.add(s2, fill_value=0)
```

In Pandas, you can add two Series using either the + operator or the <br>
.add() method. While both produce similar results, using methods offers <br>
advantages.

The .add() method integrates naturally into method chaining.

```python
city_mpg
```

```python
highway_mpg
```

```python
city_mpg.add(highway_mpg)
```

Calculate the average of city and highway mileage by using the <br>
.div(2) method.

Chaining: <br>
.add(...).div(...)

```python
city_mpg.add(highway_mpg).div(2)
```
