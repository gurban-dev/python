If you owned a restaurant, would you want a list of every order? <br>
Or, a summary of revenue, customers and averages?

Raw data is detailed, but can also be messy.
Aggregation is conducive to decision-making.

Aggregation is the process of collapsing many values to get one
meaningful number.

```python
import pandas as pd

# Create the data.
median_wages = [
        88000, 62270, 61590, 59400, 58560,
        58400, 58050, 57230, 56940, 54040,
        53810, 53390, 53260, 53020, 52610,
        52410, 51030, 50320, 50000, 49440,
        49160, 48930, 48810, 48600, 48550,
        48300, 48060, 47990, 47670, 47500,
        47360, 47020, 46950, 46930, 46860,
        46850, 46470, 46440, 46390, 46120,
        45870, 45740, 45620, 44760, 43950,
        43830, 43770, 43320, 41020, 39070
]

states =  [
        "District of Columbia", "Massachusetts", "Washington", "Alaska", "New York",
        "Connecticut", "Maryland", "New Jersey", "California", "Rhode Island",
        "Minnesota", "Oregon", "Hawaii", "Virginia", "New Hampshire",
        "Vermont", "Delaware", "North Dakota", "Illinois", "Maine",
        "Wyoming", "Wisconsin", "Arizona", "Utah", "Pennsylvania",
        "Michigan", "Ohio", "Nebraska", "Iowa", "Texas",
        "Montana", "Georgia", "North Carolina", "Indiana", "Florida",
        "Kansas", "Idaho", "Nevada", "Missouri", "Tennessee",
        "New Mexico", "Kentucky", "South Dakota", "South Carolina", "Oklahoma",
        "Alabama", "Louisiana", "West Virginia", "Arkansas", "Mississippi"
]

# Create a DataFrame.
median_wage_series = pd.Series(median_wages, index=states)

# Display the result.
print(median_wage_series)
```

<h3>Average/Mean</h3>

```python
median_wage_series.mean()
```

<h3>Largest Value</h3>

```python
median_wage_series.max()
```

<h3>Total</h3>

```python
median_wage_series.sum()
```

<h3>Quantiles</h3>

Quantiles help you understand the distribution, not just averages.

```python
# The default value for the .quantile() method is 0.5 if an argument
# is not provided.

# median_wage_series.quantile(0.5) reveals the median, which is the
# 50th percentile.

# It's the value where half of the data is below it and half is above it.
median_wage_series.quantile()
```

```python
# The top ten percent cutoff.
median_wage_series.quantile(0.9)
```

```python
# Returns multiple percentile values.
median_wage_series.quantile([0.1, 0.5, 0.9])
```
