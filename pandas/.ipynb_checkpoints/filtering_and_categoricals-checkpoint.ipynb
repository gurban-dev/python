**Categorical data** represents qualitative variables.

These are variables that can be divided into groups.

Categorical data can be split into two parts:

Nominal <br>
Favourite ice cream flavour: <r>
Vanilla <br>
Chocolate <br>
Saffron

Ordinal <br>
Socioeconomic status: <br>
High income <br>
Middle income <br>
Low income

**Quantitative data** represents variables that have a numerical component
to them.

Categoricals allow you to define logical order, which is impossible <br>
with plain strings.

```python
s = pd.Series(
    ['m', 'l', 'xs', 's', 'xl'],
    dtype='category'
)
```

```python
s
```

reorder_categories() changes the order of the categories in <br>
a categorical Series without changing the actual data values.

It makes it possible to define a meaningful order as opposed to <br>
a default alphabetical order.

The function returns a new Series with the same values, but with <br>
the categories reordered according to the list provided as an <br>
argument.

The keyword argument **ordered=True** tells Pandas to treat the <br>
catgories as ranked values rather than just labels. This way, <br>
operations like sorting and comparisons that depend on order <br>
can be performed.

```python
s.cat.reorder_categories(
    ['xs', 's', 'm', 'l', 'xl'],
    ordered=True
)
```

's' references a Pandas Series.

.cat is the categorical accessor (only works if the Series dtype is "category").

.ordered is a boolean property indicating whether the categories are ordered.

```python
s.cat.ordered
```

```python
size_type = pd.api.types.CategoricalDtype(
    categories=['xs', 's', 'm', 'l', 'xl'],
    ordered=True
)
```

```python
size_type
```

```python
s2 = pd.Series(['xs', 's', 'm', 'l', 'xl'])
```

```python
s2
```

Convert a non-categorical series to an ordered category.

```python
s3 = s2.astype(size_type)
```

```python
s3
```

The following code creates a rule for how the days of the week should be treated in the dataset.

Specifically, it tells pandas:
These are the only valid categories for this variable: <br>
Monday -> Tuesday -> Wednesday -> Thursday -> Friday -> Saturday -> Sunday

The categories have a meaningful order.

Monday comes before Tuesday, Tuesday before Wednesday, and so on.

Any column that uses this definition should follow that order, instead of being treated like ordinary text.

```python
day_type = pd.CategoricalDtype(
    categories=[
        "Monday", "Tuesday", "Wednesday",
        "Thursday", "Friday", "Saturday", "Sunday"
    ],
    ordered=True
)
```

```python
days_of_week_series = pd.Series(
    ["Wednesday", "Monday", "Friday", "Sunday"],
    dtype=day_type
)
```

```python
days_of_week_series
```

**Sort the day of the week.**

.sort_values() sorts the data in a Series or DataFrame based on the values it contains.

In this context (with days of the week defined as an ordered categorical type), it <br>
will sort the days according to the order you defined, not alphabetically.

```python
days_of_week_series.sort_values()
```
