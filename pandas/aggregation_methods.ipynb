Aggregation is the process of compressing an entire column (Series) <br>
into a single meaningful number.

Think of taking many values and reducing them to one summary value.

```python
import pandas as pd

city_mpg = pd.Series([
    18, 22, None, 17, 30,
    15, 28, 21, None, 24,
    26, 18, 20, 23, 27
], name="city_mpg")

city_mpg
```

```python
# Average/mean value.
print("city_mpg.mean():", city_mpg.mean())

# Total
print("city_mpg.sum():", city_mpg.sum())

# Highest value
print("city_mpg.max():", city_mpg.max())

# Lowest value
print("city_mpg.min():", city_mpg.min())

# Non-missing values are excluded from the count that
# is computed from counting the number of items in the
# Series.
print("city_mpg.count():", city_mpg.count())

# Number of unique values
print("city_mpg.nunique():", city_mpg.nunique())
```

```python
# Notice how unlike .count(), .size includes all rows in the Series,
# including missing values.

# .size is a property, not a method, which is why it is not followed
# by a pair of parentheses.
print("city_mpg.size:", city_mpg.size)
```
