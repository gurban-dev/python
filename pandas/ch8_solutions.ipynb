Part 1: Create a Numeric Series and Inspect Memory Usage

```python
import pandas as pd
import numpy as np

numbers = pd.Series([10, 20, 30, 40, 50])

print(numbers)
print()

# Print the dtype of the Series.
print("dtype:", numbers.dtype)

# Print the number of bytes used by the underlying data.
print("nbytes:", numbers.nbytes)
```

Part 2: Convert to Smaller Integer Types

```python
# Convert the Series to int8.
numbers_int8 = numbers.astype("int8")

# Print the dtype.
print("dtype:", numbers_int8.dtype)

# Print memory usage.
print("nbytes:", numbers_int8.nbytes)
```

Part 3: Trigger an Overflow Problem

```python
# Create a Series containing 128.
overflow_series = pd.Series([128])

# Attempt conversion to int8.
overflow_int8 = overflow_series.astype("int8")

print(overflow_int8)
```

Part 4: Use convert_dtypes()

```python
df = pd.DataFrame({
    'age': [25, 30, 35],
    'name': ['Alice', 'Bob', 'Charlie'],
    'is_student': [True, False, True]
})

print(df.dtypes)
print()

# Convert columns to pandas' newer nullable dtypes.
df_converted = df.convert_dtypes()

# Print the new dtypes.
print(df_converted.dtypes)
```
