**Exercise 3**

Write a program that:

Starts with a list of tuples, where each tuple represents a student’s name and grade.

Example:

students = [("Alice", 85), ("Bob", 90), ("Charlie", 72), ("Diana", 95)]

Converts this list into a dictionary where the student’s name is the key and the grade is the value.

Adds a new student "Eve" with a grade of 88.

Updates "Charlie"’s grade to 80.

Prints:

The student with the highest grade

The student with the lowest grade

The class average

💡 Hint: You’ll need max(), min(), and sum() with .values() for the dictionary.
