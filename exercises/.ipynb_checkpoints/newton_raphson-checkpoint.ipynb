sqrt2 is a user-defined Python function.

A function should have one responsibility.

```python
# The first line of a function declaration is known as the
# function header/signature.

"""
This is a Python docstring. Also known as a multi-line comment.

Comments are ignored by the Python interpreter.

Whereas lines of source code are instructions sent to the computer.
They are a way of communicating with a computer.
"""

# n is the parameter.
# : int is a type hint/annotation that tells whoever is reading this
# program should expect that the sqrt2() function receives an integer.
def sqrt2(n: int):
    """
    Calculate the square root of 2 using the Newton-Raphson method.

    Parameters:
    n (int): Number of accurate decimal places desired.

    Returns:
    float: Approximation of sqrt(2)

    A float is one of the data types in the Python programming language.

    A float is a number that contains a decimal point, but that is not
    surrounded by quotation marks.

    E.g. 10.0, 5., 0.5
    """

    # Step 1: Define the function f(x) = x^2 - 2
    def f(x):
        # return x^2 - 2
        # Exponentiation in Python is written by using two asterisks (**).
        return x**2 - 2

    # Step 2: Define its derivative f'(x) = 2x
    def df(x):
        # Use the "return" keyword (without quotaton marks) when a function
        # has to send data back to the point in the program where it was called.

        # Multiplication
        return 2*x

    # Step 3: Initial guess.
    x = 1.0

    # Step 4: Define the tolerance based on the number of decimal places.
    # n is the number of decimal places you want to be accurate.
    # 10**(-n) gives a number like 0.00001 if n = 5.
    # Multiplying by 0.5 ensures the stopping criterion is a bit stricter
    # than just 10^-n, which helps guarantee that the result is accurate
    # to at least n decimal places.
    tolerance = 0.5 * 10**(-n)

    # Step 5: Newton-Raphson iteration
    while True:
        # Newton-Raphson formula.
        x_new = x - f(x) / df(x)

        # abs() removes the negative from the following calculation:
        # x_new - x

        # E.g. -1.41421 -> 1.41421
        if abs(x_new - x) < tolerance:
            # Have the flow of the program exit out of the while loop.
            break

        x = x_new

    # Step 6: Return the result rounded to n decimal places.
    return round(x_new, n)        
```

Calculates sqrt(2) accurate to 5 decimal places.

```python
print(sqrt2(5))
```

1. We wanted the square root of 2.
   Mathematically, the square root of 2 is the solution to the equation:
   x^2 - 2 = 0

   x^2 = 2

   square_root(x^2) = square_root(2)

   x = square_root(2)

2. We used the Newton-Raphson method to solve this equation iteratively.

   Newton-Raphson formula:

   x_new = x - (f(x) / f'(x))


   Here, f(x) = x^2 - 2 and its derivative f'(x) = 2x.

3. We began with an initial guess (x=1.0) and repeatedly applied the formula.

   Each step brings the guess closer to the true square_root(2).


4. We stopped iterting/looping when the guess was accurate enough.

   "Accurate enough" means the difference between the two successive

   or consecutive guesses is smaller than a calculated tolerance:

   tolerance = 0.5 * 10^-n

   This ensures that the result is correct to n decimal places.

5. Finally, we rounded the result to the requested number of decimals.
