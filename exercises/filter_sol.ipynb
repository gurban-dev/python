```python
import pandas as pd
```

Create a Series that contains the temperatures for the last <br>
seven days.

```python
temperatures = pd.Series([18, 21, 19, 23, 17, 20, 22])
```

```python
temperatures
```

Compute the mean/average temperature over the last seven days <br>
by calling the .mean() method on the Series.

```python
mean_temperature = temperatures.mean()
```

```python
mean_temperature
```

Create a new Series that is the result of filtering out value <br>
that are below the mean.

```python
temps_below_mean = temperatures[temperatures < mean_temperature]
```

```python
temps_below_mean
```

Create a Series with favourite colours using a cateogrical dtype.

```python
colours = pd.Series(
    ['blue', 'green', 'red', 'blue', 'black'],
    dtype="category"
)
```

```python
colours
```

Inspect the categories by using the 'categories' attribute.

```python
colours.cat.categories
```
