<h2>Part 1</h2>

**Import Pandas**

```python
import pandas as pd
```

**Create a Series of Colours**

```python
colours_series = pd.Series(['golden orange', 'orange', 'purple'], name="Colours")
```

```python
colours_series
```

**Create a Series of Fruits**

```python
fruits_series = pd.Series(['apricots', 'persimmons', 'figs'], name="Fruits")
```

```python
fruits_series
```

**Merge them into a DataFrame**

```python
df = pd.DataFrame({
    'Colours': colours_series,
    'Fruits': fruits_series
})
```

```python
df
```

<h2>Part 2</h2>

Extracting one of the columns from the DataFrame.

```python
df["Fruits"]
```

Data type of the returned column.

```python
type(df["Fruits"])
```

Add a row to the DataFrame with a duplicate colour.

```python
# len(df) -> finds the next available row index in the DataFrame
# .loc[...] -> assigns values to that row index.
df.loc[len(df)] = ["orange", "kumquat"]
```

Inspect the updated DataFrame.

```python
df
```

Return only the rows that contain the duplicate colour.

```python
# subset='Colours' checks duplicates inside the 'Colours' column.

# keep='first' marks everything after the first duplicate.
# keep='last' marks everything before the last duplicate.
# keep=False marks everything in the duplicate group.

# df[...] filters only the rows containing the duplicates.
df[df.duplicated(subset='Colours', keep=False)]
```
