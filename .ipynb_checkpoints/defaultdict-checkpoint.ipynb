```python
from collections import defaultdict
```

 defaultdict is a dictionary that provides a default <br>
 value for keys that don't exist in the dictionary yet.

```python
# The default value for keys that don't exist yet is zero (0).
doc_freq = defaultdict(int)
```

```python
words = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
```

```python
for word in words:
    # If this was just a normal dictionary, trying to
    # access the key 'apple' would generate an error
    # because 'apple' doesn't exist yet in the dictionary.
    print('doc_freq[\'apple\']:', doc_freq['apple'])

    # doc_freq['apple'] = doc_freq['apple'] + 1
    doc_freq[word] += 1
```

```python
print(doc_freq)
```
