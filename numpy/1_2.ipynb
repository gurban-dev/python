Create a 5x4 matrix: 5 accounts × 4 asset types (Stock, Bond, ETF, Crypto)
Values represent holdings in €1000s

Questions:
1. Create the matrix with random values between 10 and 100.
2. Calculate total holdings per account (row sums).
3. Calculate total holdings per asset type (column sums).
4. Find which account has the highest total holdings.

```python
import numpy as np
```

Question 1

```python
# 10 and 100 passed as the first and second arguments
# indicate that the random integers should be between
# 10 and 100 (both included).
holdings = np.random.randint(10, 100, size=(5, 4))
```

```python
print('holdings:', holdings)
```

Question 2

```python
# The axis=1 keyword argument specifies that the sum of
# each row should be computed.
account_totals = np.sum(holdings, axis=1)
```

```python
print('account_totals:', account_totals)
```

Question 3

```python
# The axis=0 keyword argument specifies that the sum of
# each column should be computed.
asset_totals = np.sum(holdings, axis=0)
```

```python
print('asset_totals:', asset_totals)
```

Question 4

```python
# argmax() returns the index of the maximum value.
max_account_index = np.argmax(account_totals)
```

```python
print('max_account_index:', max_account_index)
```

```python
max_holdings = account_totals[max_account_index]
```

318 is the sum for the fourth row.

Keep in mind it is also the row with the maximum sum.

```python
print('max_holdings:', max_holdings)
```
