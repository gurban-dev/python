Question 1 Solution

```python
# Converts the monthly return percentages to decimal format
# and calculate compound return.
monthly_returns_decimal = monthly_returns / 100
```

```python
print(monthly_returns_decimal)
```

Suppose you begin with $100 and gain 2.5%.
The outcome is 100 * 1.025 -> 102.5

growth_factor = 1 + (monthly_returns / 100)

```python
cumulative_return = np.prod(1 + monthly_returns_decimal)
```

```python
print(cumulative_return)
```

annualised_return = (cumulative_return)^(12/n) − 1

Subtracting 1 converts the growth factor back into a percentage return.

```python
annualised_return = (cumulative_return ** (12 / len(monthly_returns)) - 1)
```

Format specifier syntax:
<variable_name>:<format_specifier>

.2f

.2 indicates that the number that <variable_name> references should
be rounded to 2 digits after the decimal.

f indicates that this is a floatng-point number.

```python
print(f'Annualised Return: {annualised_return:.2f}%')
```

What is the purpose of question 1?

If you only know how much your investment grew in 3 months, that number <br>
doesn't tell you whether you're doing well compared to someone with a full <br>
year of data.

Annualizing your return converts your performance into a standard yearly rate <br>
so you can compare apples to apples.

When you can annualize results, you’re showing you understand not just data <br>
manipulation but how to make fair, real-world comparisons.
