Why is being aware of the memory address of data important?

Answer: <br>
Being aware of the memory address of the data is important because <br>
if two arrays share the same underlying memory (as with views), modifying <br>
one will also modify the other.

If we are unaware that b points to the same memory as a, we might <br>
unintentionally overwrite or lose the original data when modifying b.

Is a view the same as a shallow copy?

Answer: <br>
A view is similar to a shallow copy, but unlike a typical shallow copy, it <br>
doesn't even copy references, it directly shares the same underlying <br>
memory buffer.
