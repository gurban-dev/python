To obtain the memory address of an object in Python,

use the built-in id() function.

Notice how the memory addresses for matrices a and b are

the same.

```python
id(a)
```

```python
print(id(b))
```

The view method creates a new array object that looks at the same data.

The .view() method creates a new array object that views the same data
as the original array.

```python
c = a.view()
```

```python
c
```

```python
c is a
```

Object c does not have the same memory address as a

because unlike object b, c is a new ndarray object.

```python
id(a) == id(c)
```

```python
type(c)
```

```python
c[0, 0] = 100
```

```python
c
```

Python's built-in hex() function converts memory addresses, <br>
which are represented internally as large integers, into a more <br>
compact and human-readable format for memory addresses.

```python
hex(id(a))
```

```python
hex(a.ctypes.data)
```
