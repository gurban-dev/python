reshape(number_of_blocks, number_of_rows, number_of_columns)

2 → number of “blocks” (outermost dimension)

3 → number of rows (middle dimension)

4 → number of columns (innermost dimension)

reshape(dim1, dim2, dim3)

changes the array to have a shape of (dim1, dim2, dim3),
where: <br>
dim1 → size of the first (outermost) dimension

dim2 → size of the second dimension (rows)

dim3 → size of the third dimension (columns)

Total elements: 2 × 3 × 4 = 24

```python
c = np.arange(24).reshape(2, 3, 4)
```

```python
c
```
