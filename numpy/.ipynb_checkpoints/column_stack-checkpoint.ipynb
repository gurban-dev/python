```python
import numpy as np
```

(1, 2, 3) is a tuple data structure and it can be passed <br>
as an argument to Numpy's array() function.

```python
a = np.array((1, 2, 3))
```

```python
b = np.array((4, 5, 6))
```

The function column_stack() stacks 1D arrays as columns into a 2D array.

Each input array becomes a column in the resulting 2D array.

The number of rows is determined by the length of the input arrays.

```python
np.column_stack((a,b))
```
