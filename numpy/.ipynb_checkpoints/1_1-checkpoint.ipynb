Create three arrays for ten transactions.

Syntax:

numpy.arange(start_index, stop_index, step_value)

If only two arguments are passed to the numpy.arange() function, <br>
the step_value parameter will be implicitly assigned
1.

```python
transaction_ids = np.arange(1, 11)
```

```python
print('transaction_ids:', transaction_ids)
```

numpy.arange() returns a numpy.ndarray.

```python
print('type(transaction_ids):', type(transaction_ids))
```
