The .mean() method can be invoked on Numpy ndarrays to compute the mean/average.

```python
# Print the mean (average) of the array 'm' rounded to 2 decimal places.
print(f"The mean/average of matrix 'm' is {m.mean():.2f}.")
```

```python
print(f"The standard deviation of matrix 'm' is {m.std():.2f}.")
```

```python
# Create an array of 100 random values between 0 and 100 (uniform distribution).
uniform_array = np.random.uniform(low=0, high=100, size=100)
```

```python
uniform_array
```

uniform_array is the ndarray of data.

95 represents the percentile.

```python
# Compute the 95th percentile of the uniform array.
p95 = np.percentile(uniform_array, 95)
```

```python
p95
```

Only see values that are in the top five percent.

```python
# Select values that are in the top 5% of the uniform array.
top_5_percent = uniform_array[uniform_array > p95]
```

```python
# Check the type of the top 5% array (it will be a NumPy ndarray).
type(top_5_percent)
```

```python
# Round the top 5% values to 2 decimal places.
top_5_percent = np.round(top_5_percent, 2)
```

```python
print(f"Top 5% of values: {top_5_percent}")
```

```python
# These are 1D ndarrays which Numpy treats as vectors for most operations.
ndarray_a = np.array([1, 2, 3, 4, 5])
ndarray_b = np.array([10, 20, 30, 40, 50])
```

The .prod() method computes 1 * 2* 3 * 4 * 5 in this context.

```python
ndarray_a.prod()
```

In this context, the dot() function performs the following: <br>
1 * 10 + 2 * 20 + 3 * 30 + 4 * 40 + 5 * 50

```python
dot_product1 = np.dot(ndarray_a, ndarray_b)
```

```python
dot_product1
```

```python
dot_product2 = ndarray_a @ ndarray_b
```

```python
dot_product2
```

```python
# Compute element-wise multiplication of 'a' and 'b'.
element_wise = ndarray_a * ndarray_b

element_wise
```

The ndarray class has a method named .sum() which will simply compute the sum <br>
of all numbers.

```python
ndarray_a.sum()
```

```python
# Get the number of dimensions of 'a' (should be 1 for 1D array).
ndarray_a.ndim
```

The ndarray class has a .reshape(no_of_rows, no_of_columns) method.

```python
ndarray_a = ndarray_a.reshape(5, 1)
```

```python
# Check the number of dimensions again (should now be 2).
ndarray_a.ndim
```

```python
# Compute the cumulative sum of 'a' (running total).
ndarray_a.cumsum()
```

```python
# Generate the Fibonacci sequence up to N terms.
N = 10

# Initialise the Fibonacci sequence with the first two numbers.
fib = np.array([0, 1])

for i in range(2, N):
    # Compute the next Fibonacci number as the sum of the last two.
    next_value = fib[-1] + fib[-2]

    # Append the next value to the sequence.
    fib = np.append(fib, next_value)

# Print the generated Fibonacci sequence.
print(fib)
```
