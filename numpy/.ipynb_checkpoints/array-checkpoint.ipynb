The code demonstrates how the memory layout (C-order vs F-order)

of a NumPy array affects computational performance when performing

operations along different axes.

```python
import time

# Create a large 2D array
shape = (5000, 5000)

# Create a C-order (row-major) array
c_array = np.arange(shape[0] * shape[1], dtype=np.int32).reshape(shape, order='C')

# Create an F-order (column-major) array
f_array = np.arange(shape[0] * shape[1], dtype=np.int32).reshape(shape, order='F')

# Test performance for row-wise sum (fast for C-order)
print("Timing row-wise sum:")
start_time = time.time()
np.sum(c_array, axis=1) # Summing along rows is fast for C-order
print(f"C-order time: {time.time() - start_time:.4f}s")

start_time = time.time()
np.sum(f_array, axis=1) # Summing along rows is slower for F-order
print(f"F-order time: {time.time() - start_time:.4f}s")
print("-" * 20)

# Test performance for column-wise sum (fast for F-order)
print("Timing column-wise sum:")
start_time = time.time()
np.sum(c_array, axis=0) # Summing along columns is slow for C-order
print(f"C-order time: {time.time() - start_time:.4f}s")

start_time = time.time()
np.sum(f_array, axis=0) # Summing along columns is fast for F-order
print(f"F-order time: {time.time() - start_time:.4f}s")
```
