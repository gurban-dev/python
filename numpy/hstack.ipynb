```python
import numpy as np
```

```python
one_dim_ndarray1 = np.array((1, 2, 3))
```

```python
one_dim_ndarray2 = np.array((4, 5, 6))
```

For 1D arrays, it simply concatenates arrays by combining their items into one ndarray.

```python
np.hstack((one_dim_ndarray1, one_dim_ndarray2))
```

For 2D ndarrays, np.hstack() makes each input ndarray become a column inside of a larger two-dimensional ndarray.

```python
two_dim_ndarray1 = np.array([[1],[2],[3]])
```

```python
two_dim_ndarray2 = np.array([[4],[5],[6]])
```

```python
np.hstack((two_dim_ndarray1, two_dim_ndarray2))
```
