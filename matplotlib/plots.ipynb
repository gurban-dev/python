```python
# range(start_value, stop_value (exclusive), step_value)

# When one argument is passed to the range() function, that
# argument becomes the stop value.

# The start value defaults to 0 (zero) and the step value defaults
# to 1.
nums = range(6)

nums_lst = list(nums)

nums_lst
```

```python
# [num**2 for num in nums] iterates through each number
# in nums and squares them:
# [0, 1, 4, 9, 16, 25]

# plt.plot(horizontal_points, vertical_points)
plt.plot(nums, [num**2 for num in nums])

plt.show()
```

range(start_value, stop_value(exclusive), step_value) is Python's built-in <br>
function that generates a sequence of integers from start_value to stop_value <br>
with stop_value being excluded with an increment of step_value.

Both the start_value and step_value are optional. Arguments do not have to be <br>
provided for these parameters.

arange(x, y, z) is part of Numpy, and it generates a sequence of elements (with <br>
the data type determined by parameters types) from x to y with a spacing of z.

One difference between range() and arange() is that arange() accepts floats <br>
whereas range() does not. 'x' and 'z' are optional in the sense that they do <br>
not have to be provided as arguments.

```python
import numpy as np

# np.arange(x, y, spacing)
floats = np.arange(0.0, 6.0, 0.01)

# By using a much smaller spacing value, the smoothness of the graph
# can be enhanced.

floats
```

```python
# plt.plot(horizontal_values, vertica_values)
plt.plot(floats, [x**2 for x in floats])
```

```python
# Matplotlib gives us the ability to plot multiple lines on the
# same figure.

# The subsequent range of numbers will be utilised to plot
# three different lines.
nums = range(1, 5)

# Plot three different lines using the same data:
plt.plot(nums, [num * 1.5 for num in nums])

plt.plot(nums, [num * 3.0 for num in nums])

plt.plot(nums, [num / 1.5 for num in nums])

plt.show()
```
